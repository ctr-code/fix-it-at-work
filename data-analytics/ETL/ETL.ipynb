## Step 5: Data Quality Assessment

Before cleaning the dataset, we will assess its quality.

### Objectives
- Identify missing values.
- Detect duplicate records.
- Examine data types.
- Identify potential data quality issues.

This assessment will help us determine the appropriate cleaning and transformation steps.

```python
# Check dataset dimensions
print("Dataset Shape:")
print(df.shape)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum().to_string())

# Check duplicate rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# Check duplicate ticket references
print("\nDuplicate Ticket References:")
print(df["Ticket Ref"].duplicated().sum())

# Check data types
print("\nData Types:")
print(df.dtypes.to_string())
```

## Step 6: Missing Values Analysis

In this step, we will identify columns containing missing values and calculate the percentage of missing data.

This analysis will help us distinguish between expected missing values and potential data quality issues before applying cleaning rules.

```python
# Calculate missing values
missing_count = df.isnull().sum()

# Calculate missing percentages
missing_percentage = (missing_count / len(df)) * 100

# Create a summary table
missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage (%)": missing_percentage.round(2)
})

# Display only columns with missing values
missing_summary = missing_summary[
    missing_summary["Missing Count"] > 0
]

# Sort by highest missing count
missing_summary = missing_summary.sort_values(
    by="Missing Count",
    ascending=False
)

display(missing_summary)
```

## Step 7: Duplicate Records Analysis

In this step, we will identify duplicate records and duplicate ticket references.

### Objectives
- Check for completely duplicated rows.
- Identify duplicate ticket references.
- Assess whether duplicate records require further investigation.

No records will be removed at this stage.

```python
# Check completely duplicated rows
duplicate_rows = df.duplicated().sum()

print("Total Duplicate Rows:", duplicate_rows)

# Check duplicate ticket references
duplicate_tickets = df["Ticket Ref"].duplicated().sum()

print("Duplicate Ticket References:", duplicate_tickets)

# Display records with duplicate ticket references
if duplicate_tickets > 0:
    display(
        df[df.duplicated(subset=["Ticket Ref"], keep=False)]
        .sort_values("Ticket Ref")
    )
else:
    print("No duplicate ticket references found.")
```

## Step 8: Missing Values by Ticket Status

In this step, we will investigate missing values based on ticket status.

### Objectives
- Identify the different ticket statuses.
- Examine missing closure and resolution dates.
- Determine whether missing values are expected or require cleaning.

No missing values will be modified at this stage.

```python
# Check ticket status distribution
print("Ticket Status Distribution:")
display(df["Status"].value_counts(dropna=False))

# Check missing dates by ticket status
missing_by_status = df.groupby("Status").agg(
    Total_Tickets=("Ticket Ref", "count"),
    Missing_Date_Closed=("Date Closed", lambda x: x.isna().sum()),
    Missing_Date_Resolved=("Date Resolved", lambda x: x.isna().sum())
)

print("\nMissing Values by Ticket Status:")
display(missing_by_status)
```

## Step 9: Data Types Validation

In this step, we will examine the data types of the dataset.

### Objectives
- Identify numerical, categorical, and datetime columns.
- Check whether date columns have appropriate data types.
- Identify columns that may require type conversion.
- Prepare the dataset for data cleaning and transformation.

```python
# Step 9: Data Types Validation

# Display data types for all columns
print("Dataset Data Types:")
display(df.dtypes.to_frame(name="Data Type"))

# Identify numerical columns
numerical_columns = df.select_dtypes(include=["number"]).columns.tolist()

print("\nNumerical Columns:")
print(numerical_columns)

# Identify text columns
text_columns = df.select_dtypes(include=["object", "string"]).columns.tolist()

print("\nText Columns:")
print(text_columns)

# Identify datetime columns
datetime_columns = df.select_dtypes(include=["datetime"]).columns.tolist()

print("\nDatetime Columns:")
print(datetime_columns)
```

## Step 10: Data Consistency Validation

In this step, we will identify invalid numerical values.

### Objectives
- Check for negative response and resolution times.
- Identify negative costs.
- Validate satisfaction scores between 1 and 5.
- Check for invalid reopening counts.

No data will be modified at this stage.

```python
# Step 10: Data Consistency Validation

# Check for negative numerical values
print("Negative Response Times:",
      (df["Response Time (hrs)"] < 0).sum())

print("Negative Resolution Times:",
      (df["Resolution Time (hrs)"] < 0).sum())

print("Negative Costs:",
      (df["Cost (£)"] < 0).sum())

print("Negative Reopening Counts:",
      (df["Times Reopened"] < 0).sum())

# Check invalid satisfaction scores
invalid_satisfaction = (
    df["Satisfaction (1-5)"].notna()
    & ~df["Satisfaction (1-5)"].between(1, 5)
)

print("Invalid Satisfaction Scores:",
      invalid_satisfaction.sum())
```

## Step 11: Missing Values Treatment

In this step, we will prepare a clean copy of the dataset
and review missing values.

### Objectives
- Preserve the original dataset.
- Identify missing values that may be expected.
- Avoid replacing missing dates with artificial values.
- Prepare the data for further cleaning.

Missing values will only be modified when there is a clear
business justification.

```python
# Step 11: Missing Values Treatment

# Create a copy to preserve the original dataset
df_clean = df.copy()

# Identify columns containing missing values
missing_columns = df_clean.isna().sum()

missing_columns = missing_columns[missing_columns > 0]

print("Columns with Missing Values:")
display(missing_columns.to_frame(name="Missing Count"))

# Check dataset size after creating the clean copy
print("\nOriginal Dataset Shape:", df.shape)
print("Clean Dataset Shape:", df_clean.shape)
```

## Step 12: Validate Missing Values Against Ticket Status

### Objectives
- Check whether missing dates are consistent with ticket status.
- Identify closed tickets without a closing date.
- Identify resolved tickets without a resolution date.
- Preserve legitimate missing values.

```python
# Step 12: Validate Missing Values Against Ticket Status

# Check closed tickets without a closing date
closed_missing_date = df_clean[
    (df_clean["Status"] == "Closed") &
    (df_clean["Date Closed"].isna())
]

# Check resolved tickets without a resolution date
resolved_missing_date = df_clean[
    (df_clean["Status"] == "Resolved") &
    (df_clean["Date Resolved"].isna())
]

print("Closed tickets missing Date Closed:",
      len(closed_missing_date))

print("Resolved tickets missing Date Resolved:",
      len(resolved_missing_date))

print("\nDataset Shape:", df_clean.shape)
```

## Step 13: Validate Date Chronology

### Objectives
- Check the chronological order of ticket dates.
- Identify invalid date sequences.
- Ensure data reliability for SLA analysis.

```python
# Step 13: Validate Date Chronology

# First response before reporting date
invalid_first_response = (
    df_clean["First Response"] < df_clean["Date Reported"]
).sum()

# Resolution before reporting date
invalid_resolution = (
    df_clean["Date Resolved"] < df_clean["Date Reported"]
).sum()

# Closing before resolution date
invalid_closing = (
    df_clean["Date Closed"] < df_clean["Date Resolved"]
).sum()

print("Invalid First Response Dates:", invalid_first_response)
print("Invalid Resolution Dates:", invalid_resolution)
print("Invalid Closing Dates:", invalid_closing)

print("\nDataset Shape:", df_clean.shape)
```

## Step 14: SLA Consistency Validation

### Objectives
- Inspect response and resolution SLA values.
- Identify missing or unexpected SLA labels.
- Prepare SLA data for Power BI analysis.

```python
# Step 14: SLA Consistency Validation

# Check Response SLA values
print("Response SLA Distribution:")
print(df_clean["Response SLA"].value_counts(dropna=False))

# Check Resolution SLA values
print("\nResolution SLA Distribution:")
print(df_clean["Resolution SLA"].value_counts(dropna=False))

# Check missing SLA values
print("\nMissing Response SLA:",
      df_clean["Response SLA"].isna().sum())

print("Missing Resolution SLA:",
      df_clean["Resolution SLA"].isna().sum())

# Check dataset shape
print("\nDataset Shape:", df_clean.shape)
```

## Step 15: Inspect SLA Settings

### Objectives
- Load the SLA Settings worksheet.
- Identify response and resolution time targets.
- Prepare for SLA validation.

```python
# Step 15: Inspect SLA Settings

# Load SLA settings
sla_settings = pd.read_excel(
    excel_path,
    sheet_name="SLA Settings"
)

# Display SLA settings
print("SLA Settings Shape:", sla_settings.shape)

print("\nSLA Settings Columns:")
print(sla_settings.columns.tolist())

print("\nSLA Settings Data:")
display(sla_settings)
```

## Step 16: Validate SLA Against Priority Targets

### Objectives
- Clean the SLA settings table.
- Match SLA targets with ticket priorities.
- Validate missing SLA targets.
- Prepare data for Power BI.

```python
# Step 16: Validate SLA Against Priority Targets

# Keep valid SLA priorities
valid_priorities = ["Urgent", "High", "Medium", "Low"]

sla_clean = sla_settings[
    sla_settings["Priority"].isin(valid_priorities)
].copy()

# Merge SLA targets with the cleaned dataset
df_clean = df_clean.merge(
    sla_clean,
    on="Priority",
    how="left",
    validate="many_to_one"
)

# Check missing SLA targets
print("Missing Response Targets:",
      df_clean["Response Target (hrs)"].isna().sum())

print("Missing Resolution Targets:",
      df_clean["Resolution Target (hrs)"].isna().sum())

# Display sample records
display(
    df_clean[
        [
            "Ticket Ref",
            "Priority",
            "Response Time (hrs)",
            "Response Target (hrs)",
            "Resolution Time (hrs)",
            "Resolution Target (hrs)"
        ]
    ].head()
)

print("\nDataset Shape:", df_clean.shape)
```

## Step 17: Export Cleaned Dataset

### Objectives
- Export the cleaned dataset.
- Preserve the original raw data.
- Prepare the final dataset for EDA and Power BI.

```python
# Step 17: Export Cleaned Dataset

from pathlib import Path

# Define output folder
output_folder = Path.cwd().parent / "Data" / "Processed"

# Create folder if it does not exist
output_folder.mkdir(parents=True, exist_ok=True)

# Define output file
output_file = output_folder / "cleaned_tickets.csv"

# Export cleaned dataset
df_clean.to_csv(output_file, index=False, encoding="utf-8-sig")

# Verify exported file
exported_df = pd.read_csv(output_file)

print("Export successful!")
print("File saved to:", output_file)
print("Exported Dataset Shape:", exported_df.shape)
print("File exists:", output_file.exists())
```
